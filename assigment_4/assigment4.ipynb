```python
import pandas as pd

roll_number = "1024170320"

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

last_two = roll_number[-2:]

categories = ["billing", "account", "general"]

personalized_entries = []

for digit in last_two:
    d = int(digit)
    category = categories[d % 3]

    if category == "billing":
        entry = {
            "question": "how can i check my payment status",
            "answer": "You can check your payment status from the billing section.",
            "keywords": "payment status billing transaction",
            "category": category
        }
    else:
        entry = {
            "question": "how do i update my registered mobile number",
            "answer": "You can update your registered mobile number from account settings.",
            "keywords": "mobile number update account",
            "category": category
        }

    personalized_entries.append(entry)

final_entries = fixed_entries + personalized_entries

df = pd.DataFrame(final_entries)

print(df)
```

```python

def score_query(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keywords))

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return results


query = input("Enter your query: ")

matches = score_query(query, df)

for result in matches:
    print("Question:", result["question"])
    print("Answer:", result["answer"])
    print("Category:", result["category"])
    print("Confidence Score:", result["score"])
    print("-------------------")
```

```python
def score_query(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keywords))

        results.append({
            "question": row["question"],
            "answer": row["answer"],
            "category": row["category"],
            "score": score
        })


    max_score = max(item["score"] for item in results)


    best_matches = [
        item for item in results
        if item["score"] == max_score
    ]


    if len(best_matches) > 1:
        print("Tie found. Multiple matching entries:\n")

    else:
        print("Best match:\n")


    for item in best_matches:
        print("Question:", item["question"])
        print("Answer:", item["answer"])
        print("Category:", item["category"])
        print("Score:", item["score"])
        print("----------------")


print("Tie Query:")
score_query("fee", df)


print("\nNon-Tie Query:")
score_query("password", df)
```
