```python
n = int(input("Enter n: "))

answer = 0

for i in range(2, n + 1):

    is_prime = True

    for j in range(2, int(i ** 0.5) + 1):
        if i % j == 0:
            is_prime = False
            break

    if is_prime:
        answer += i

print("Sum of prime numbers =", answer)
```

```python
def check_prime(num):

    if num < 2:
        return False

    for factor in range(2, int(num ** 0.5) + 1):
        if num % factor == 0:
            return False

    return True


print(check_prime(17))
print(check_prime(18))
print(check_prime(41))
print(check_prime(100))
```
