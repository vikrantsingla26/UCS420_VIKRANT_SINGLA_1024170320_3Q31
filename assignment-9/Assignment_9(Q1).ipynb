```python

import os
import io
import numpy as np
import easyocr

from PIL import Image
from dotenv import load_dotenv
from IPython.display import display, Markdown, Audio
from gtts import gTTS
from landingai.predict import Predictor
from tkinter import Tk, filedialog


# =====================================
# 1. LOAD ENVIRONMENT VARIABLES
# =====================================

load_dotenv()

API_KEY = os.getenv("LANDINGAI_API_KEY")

if not API_KEY:
    raise ValueError("LandingAI API key not found in .env file")


# =====================================
# 2. LANDINGAI CONFIGURATION
# =====================================

ENDPOINT_ID = "YOUR_VALID_DEPLOYED_ENDPOINT_ID"

predictor = Predictor(
    endpoint_id=ENDPOINT_ID,
    api_key=API_KEY
)

print("LandingAI predictor initialized.")


# =====================================
# 3. INITIALIZE OCR
# =====================================

reader = easyocr.Reader(['en'])

print("EasyOCR initialized.")


# =====================================
# 4. MEDICINE LABEL ASSISTANT
# =====================================

def medicine_label_assistant():

    print("\nMEDICINE LABEL READING COGNITIVE ASSISTANT")
    print("Select an image of a medicine bottle.")

    # Select image from Mac
    root = Tk()
    root.withdraw()
    root.attributes('-topmost', True)

    image_path = filedialog.askopenfilename(
        title="Select Medicine Label Image",
        filetypes=[
            ("Image files", "*.jpg *.jpeg *.png")
        ]
    )

    root.destroy()

    if not image_path:
        print("No image selected.")
        return

    # =================================
    # 5. LOAD IMAGE
    # =================================

    with Image.open(image_path) as img:
        image = img.convert("RGB").copy()

    display(image)

    image_array = np.array(image)

    # =================================
    # 6. LANDINGAI PREDICTION
    # =================================

    print("\nRunning LandingAI model...")

    try:
        predictions = predictor.predict(image_array)

        print("\nLandingAI Model Predictions:")
        print(predictions)

    except Exception as error:
        print("LandingAI prediction failed:")
        print(error)

    # =================================
    # 7. OCR TEXT EXTRACTION
    # =================================

    print("\nExtracting medicine label text...")

    results = reader.readtext(image_array)

    extracted_text = []

    for detection in results:

        text = detection[1]
        confidence = detection[2]

        if confidence > 0.25:
            extracted_text.append(text)

    # =================================
    # 8. DISPLAY INFORMATION
    # =================================

    if not extracted_text:
        print("No readable text found.")
        return

    medicine_information = "\n".join(extracted_text)

    print("\nExtracted Medicine Label Information:")
    print(medicine_information)

    display(
        Markdown(
            "### Medicine Label Information\n\n"
            + "\n".join(
                "- " + text for text in extracted_text
            )
        )
    )

    # =================================
    # 9. GENERATE AUDIO
    # =================================

    print("\nGenerating audio response...")

    try:

        speech = gTTS(
            text=(
                "The extracted medicine label information is: "
                + medicine_information
            ),
            lang="en"
        )

        audio_path = "medicine_information.mp3"

        speech.save(audio_path)

        display(Audio(audio_path, autoplay=False))

        print("Audio saved:", audio_path)

    except Exception as error:
        print("Audio generation failed:", error)


# =====================================
# 10. RUN ASSISTANT
# =====================================

medicine_label_assistant()
```
