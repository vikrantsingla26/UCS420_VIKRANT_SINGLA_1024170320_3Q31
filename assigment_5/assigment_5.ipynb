```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print( most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

Vikrant = np.linspace(10, 100, 25)

print("Array:")
print(Vikrant)

print("Dimensions:", Vikrant.ndim)
print("Shape:", Vikrant.shape)
print("Total elements:", Vikrant.size)
print(" type:", Vikrant.dtype)
print(" bytes:", Vikrant.nbytes)

transpose = Vikrant.reshape(25, 1)


print(transpose)


print(Vikrant.T)
```

```python
import numpy as np

ucs420_Vikrant = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])


print(ucs420_Vikrant)

print("Mean:", np.mean(ucs420_Vikrant))
print("Median:", np.median(ucs420_Vikrant))
print("Maximum:", np.max(ucs420_Vikrant))
print("Minimum:", np.min(ucs420_Vikrant))
print("Unique elements:", np.unique(ucs420_Vikrant))

reshaped_ucs420_Vikrant = ucs420_Vikrant.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_Vikrant)

resized_ucs420_Vikrant = np.resize(ucs420_Vikrant, (2, 3))

print("Resized array:")
print(resized_ucs420_Vikrant)
```
